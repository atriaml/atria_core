```python
from pathlib import Path

from atria_core.logger import get_logger

logger = get_logger(__name__)


test_sample_path = "/home/saifullah/test_sample.pdf"
```

```python
from atria_core.types import PDF, DocumentContent, DocumentInstance, Image, TextElement
from atria_core.types._generic._bounding_box import BoundingBox


def from_pdf(file_path: Path):
    return DocumentInstance(
        sample_id=file_path.stem,
        page_id=0,
        pdf=PDF(
            file_path=str(file_path),
            num_pages=1,
        ),
    )


def from_file(file_path: str | Path):
    file_path = Path(test_sample_path)
    assert Path(test_sample_path).exists(), f"File `{file_path}` does not exist."

    # check for the file extension
    file_extension = file_path.suffixes[-1]
    if file_extension == ".pdf":
        return from_pdf(file_path=file_path)
    elif file_extension in [".png", ".jpeg", ".jpg"]:
        return from_pdf(file_path=file_path)
    else:
        raise RuntimeError(f"Unsupported file extension `{file_extension}`")

doc = from_file(file_path=test_sample_path)
```
