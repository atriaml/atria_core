```python
from pathlib import Path

from atria_core.logger import get_logger

logger = get_logger(__name__)


test_sample_path = "/home/saifullah/test_image.jpeg"
```

```python
import time

from atria_core.types._data_instance._image_instance import ImageInstance

image_instance = ImageInstance(file_path=test_sample_path)

# First access (loads from disk)
start = time.perf_counter()
_ = image_instance.image
first_time = time.perf_counter() - start

# Cached access (repeat many times)
N = 100_000

start = time.perf_counter()
for _ in range(N):
    _ = image_instance.image
cached_total = time.perf_counter() - start

print(f"First load:        {first_time:.6f} s")
print(f"Cached total:      {cached_total:.6f} s ({N:,} accesses)")
print(f"Cached avg/access: {cached_total / N * 1e6:.3f} µs")
```
